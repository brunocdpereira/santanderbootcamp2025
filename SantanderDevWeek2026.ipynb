```python
import pandas as pd
import google.generativeai as genai
import time


# CONFIGURAÇÃO DA API
genai.configure(api_key="removed")
model = genai.GenerativeModel('gemini-1.5-flash')

```

```python
# ETAPA 1: EXTRACT

print(" ETAPA 1: EXTRACT")

# Lê o arquivo CSV
df = pd.read_csv('SDW2025.csv')

# Converte para lista de dicionários
users = df.to_dict(orient='records')

# Inicializa campo 'news' vazio
for user in users:
    user['news'] = []

print(f" {len(users)} usuários carregados.\n")
```

```python
# ETAPA 2: TRANSFORM

print("ETAPA 2: TRANSFORM")

for i, user in enumerate(users, 1):
    print(f"   [{i}/{len(users)}] Processando: {user['Name']}...", end="\r")

    prompt = f"Crie uma mensagem curta de marketing sobre investimentos para {user['Name']} (máximo 100 caracteres)"

    try:
        response = model.generate_content(prompt)
        message = response.text.strip()

        if not message:
            message = f"{user['Name']}, invista no seu futuro com o Santander"
    except:
        message = f"{user['Name']}, venha investir conosco em 2026."

    user['news'].append({
        "icon": "https://github.com/digitalinnovationone/santander-dev-week-2023-api/blob/main/docs/icons/credit.svg",
        "description": message[:100]
    })

    time.sleep(1.5)

print("\n Mensagens geradas com sucesso!\n")

```

```python
# ETAPA 3: LOAD


print("ETAPA 3: LOAD\n")

# Cria DataFrame final
df_final = pd.DataFrame([{
    'UserID': u['UserID'],
    'Name': u['Name'],
    'Account': u['Account'],
    'Mensagem_Marketing': u['news'][0]['description']
} for u in users])

# Salva em CSV
output_file = 'campanha_marketing_santander.csv'
df_final.to_csv(output_file, index=False, encoding='utf-8')

print(f"Arquivo '{output_file}' salvo com sucesso!\n")


# EXIBE RESULTADO


print("RESULTADO FINAL:")
print(df_final)

# DOWNLOAD DO ARQUIVO

from google.colab import files
files.download(output_file)

print(f"\n Download de '{output_file}' iniciado!\n")


print("="*60)
print("Dev by: Bruno Pereira")
print("LinkedIn: https://www.linkedin.com/in/brunocdpereira/")
print("="*60)
```
